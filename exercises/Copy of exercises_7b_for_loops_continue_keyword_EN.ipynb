14)
Ask the user to enter two numbers: a start and a stop.

If start < stop → print all numbers from start to stop counting up.

If start > stop → print all numbers from start to stop counting down.

```python
start = int(input("enter a number: "))
stop = int(input("enter a second number: "))

if start < stop:
    for i in range (start, stop+1):
        print(i)
else:
    for i in range (start, stop-1,-1):
        print(i)
```
