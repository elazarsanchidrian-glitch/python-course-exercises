17)
Start with this list:

colors = ["blue", "green", "yellow", "purple", "orange", "red"]

Write code that loops through this list

Then use an if-else conditional that checks the length of a the color.

If the word is more than 5 letters then print "The word is long, over 5 letters (add the color)"

If the word is more than 5 letters then print "The word is short, under 5 letters (add the color)"


```python
colors = ["blue", "green", "yellow", "purple", "orange", "red"]
for color in colors:
    print(color)
    if len(color)>5:
        print("longer than 5")
    if len(color)<5:
        print("shorter than 5")
```
