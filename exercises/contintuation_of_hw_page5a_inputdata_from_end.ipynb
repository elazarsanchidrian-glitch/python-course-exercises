```python
number_1 = 4
text_1 = "sun"
decimal_num_1 = 2.0
decimal_num_2 = 4.4

She’s asking you to practice how the * operator works differently depending on the data types in Python.

Here’s the exercise explained step by step.

# 4 variables
number_1 = 4              # int
text_1 = "sun"            # str
decimal_num_1 = 2.0       # float
decimal_num_2 = 4.4       # float


# a) int * float
result_a = number_1 * decimal_num_1
print(result_a)

# Explanation:
# 4 * 2.0 = 8.0
# Python gives a float result


# b) float * float
result_b = decimal_num_1 * decimal_num_2
print(result_b)

# Explanation:
# 2.0 * 4.4 = 8.8


# c) int * str
result_c = number_1 * text_1
print(result_c)

# Explanation:
# The string repeats 4 times
# Output: sunsunsunsun


# d) float * str
result_d = decimal_num_1 * text_1
print(result_d)

# Explanation:
# This causes an ERROR
# Python cannot multiply a string by a float

What happens in each case:

int * float
Python does normal multiplication.
Result becomes a float.
float * float
Regular decimal multiplication.
int * str
Python repeats the string that many times.
float * str
Error happens because strings can only be multiplied by integers, not floats.

The error looks like this:

TypeError: can't multiply sequence by non-int of type 'float'
```

```python
When you replace * with +, the operator behaves differently depending on the data types.

Here’s the updated code and explanation:

# 4 variables
number_1 = 4              # int
text_1 = "sun"            # str
decimal_num_1 = 2.0       # float
decimal_num_2 = 4.4       # float


# a) int + float
result_a = number_1 + decimal_num_1
print(result_a)

# Explanation:
# 4 + 2.0 = 6.0
# Python converts the int into a float


# b) float + float
result_b = decimal_num_1 + decimal_num_2
print(result_b)

# Explanation:
# 2.0 + 4.4 = 6.4


# c) int + str
result_c = number_1 + text_1
print(result_c)

# Explanation:
# ERROR
# Python cannot add a number and a string directly


# d) float + str
result_d = decimal_num_1 + text_1
print(result_d)

# Explanation:
# ERROR
# Python cannot add a float and a string

What changed compared to *:

Operation	With *	With +
int + float	Mathematical multiplication	Mathematical addition
float + float	Multiplication	Addition
int + str	Repeats string	Error
float + str	Error	Error

Important difference:

* can repeat strings when used with an integer:
3 * "hi"
# Output: hihihi
+ joins (concatenates) strings only with other strings:
"hi" + "bye"
# Output: hibye

But:

4 + "hi"

causes:

TypeError: unsupported operand type(s) for +: 'int' and 'str'
```
