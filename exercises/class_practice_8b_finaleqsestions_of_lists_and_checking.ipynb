```python
import random

prisoner_list = ["khajiit", "argonian", "bosmer"]

while prisoner_list:

    dead_prisoner = random.choice(prisoner_list)
    prisoner_list.remove(dead_prisoner)

    print(dead_prisoner, "goes to the block!")

    random.shuffle(prisoner_list)

    print("Survivors:")
    for prisoner in prisoner_list:
        print("-", prisoner)

    print()
```
