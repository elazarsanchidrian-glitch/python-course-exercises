5) Write code to add numbers until total up to 100

1. Use a while loop to add numbers until the total is up to 100 and not more.

2. Inside the loop use an input() to ask the user for a whole number

3. Keep track of the sum of all of the numbers that the user enters.

4. Test the code with a few sets of values.


```python
total_sum = 0

while True:
    number = int(input("enter a whole number: "))

    if total_sum + number > 100:
        break

    total_sum = total_sum + number

print(total_sum)
```

```python
sum_user_number = 0
user_number = 0

while sum_user_number < 100:
  user_number = int(input("enter a number: "))
  print("user_number", user_number)

  if sum_user_number + user_number > 100:
    break
  else:
    sum_user_number = user_number + sum_user_number
    print("sum_user_number", sum_user_number)

print("final sum: ", sum_user_number)
```

6) "7 boom" -

Write a program that receives an integer from the user as input.
The program will use a while loop
and will print all the numbers from 1 to the number the user entered,
but when it reaches a number divisible by 7 it will print "boom".
For example, for the number 16 we will print: 1 2 3 4 5 6 Boom 8 9 10 11 12 13 Boom 15 16
(hint - use the % operator)

```python
number = int(input("enter a number: "))
while number <= 10:
  print(number)
  number = number + 1
  if number % 7 == 0:
    print("boom")
```

```python
number = int(input("Enter a number: "))

i = 1

while i <= number:
    if i % 7 == 0:
        print("Boom")
    else:
        print(i)

    i = i + 1
```

8)
Write a program that uses a while loop:  `while True:`
to ask the user to enter a password until they type the correct one (e.g., 'python123').
When the password is correct, print 'Access granted' and exit the loop.

```python
while True:
    password = input("enter password: ")

    if password == "python123":
        print("correct password")
        break
```

```python
count = 1

while count <= 3:
    name = input("student name: ")
    grade = int(input("student grade: "))

    print(name, grade)

    count = count + 1
```
